## Discrete Distributions

1. Probability of getting a 3 by rolling a fair die.
2. Probability of getting less than 4 by rolling a fair die.

```python
from scipy.stats import randint

low, high = 1, 7
die = randint(low, high)

print(die.pmf(3))
print(die.cdf(3))
```

1. Probability of getting 3 heads in 5 flips of a coin.
2. Probability of getting 4-5 heads in 5 flips of a coin.

```python
from scipy.stats import binom

n = 5  
p = 0.5   
coin = binom(n, p)

print(coin.pmf(3))
print(1-coin.cdf(3))
```

The average number of patients getting into the emergency room is 8 per hour.

1. Probability of 5 patients arriving in an hour.
2. Probability of 5-7 patients arriving in an hour.

```python
from scipy.stats import poisson

mu = 8  # average rate (λ)
patient= poisson(mu)

print(patient.pmf(5))
print(patient.cdf(7)-patient.cdf(5))
print(patient.pmf(6)+patient.pmf(7))
```

## Continuous Distribution

A random variable x is known to be uniformly distributed between 0 and 10.

1. Compute P(x<5)
2. Compute P(5<=x<=7)

```python
from scipy.stats import uniform

a = 0
b = 10

rv = uniform(loc=a, scale=b-a)

print(rv.pdf(5))
print(rv.cdf(5))
print(rv.cdf(7)-rv.cdf(5))
```

The national average height for male is 170cm with a standard deviation of 5cm.

1. Compute the probability of a male's height is 175cm or less.
2. Compute the probability of a male's height is more than 175cm.

```python
from scipy.stats import norm

mean = 170  
sigma = 5 
  
height = norm(loc=mean, scale=sigma)

print(height.cdf(175)) 
print(1-height.cdf(175))
```

The time between arrivals of vehicles at a particular intersection follows an exponential probability distribution with a mean of 12 seconds.

1. what is the probability that the arrival time between vehicles is 6 seconds or less?
2. What is the probability of 30 or more seconds between vehicle arrival?

```python
from scipy.stats import expon

lamda = 12  

vehicle= expon(scale=lamda)

print(vehicle.cdf(6))
print(1-vehicle.cdf(30))
```

## Exercise: Website Traffic

Assume that the traffic to the web site of Smiley’s People, Inc., which sells customized T-shirts, follows a normal distribution, with a mean of 4.5 million visitors per day and a standard deviation of 820,000 visitors per day. 

1.	What is the probability that the web site has fewer than 5 million visitors in a single day?
1.	What is the probability that the web site has 3 million or more visitors in a single day?
1.	What is the probability that the web site has between 3 million and 4 million visitors in a single day?
1.	Assume that 85% of the time, the Smiley’s People web servers can handle the daily web traffic volume without purchasing additional server capacity. What is the amount of web traffic that will require Smiley’s People to purchase additional server capacity? **HINT:** use the percent point function (ppf) of the normal distribution to find the value corresponding to the 85th percentile.

```python
from scipy.stats import norm

mean = 4500000
sigma = 820000
visitors = norm(loc=mean, scale=sigma)
print(visitors.cdf(5000000))
print(1-visitors.cdf(3000000))
print(visitors.cdf(4000000)+(1-visitors.cdf(3000000)-1))
print(visitors.ppf(.85))
```

## Plot Probability Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Poisson Distribution

```python
# Create an array of possible patient counts (the x-axis values).
# We'll go from 0 up to about twice the lambda value to see the full shape.
x1 = np.arange(0, 16)

# Calculate the probability mass function (PMF) for each x-value.
# This gives the probability of observing exactly 'x' patients.
poisson_pmf = patient.pmf(x1)

# Create the plot. A bar chart is appropriate for discrete distributions.
plt.figure(figsize=(10, 6))
plt.bar(x1, poisson_pmf, color='skyblue', edgecolor='black')

# Add labels and a title for clarity.
plt.title(f'Poisson Distribution ($\lambda$ = {mu})')
plt.xlabel('Number of Patients')
plt.ylabel('Probability')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(x1)
plt.show()
```

### Exponential Distribution

```python
# Generate x-values for the plot
# We'll go from 0 to 60 seconds to show the decay, as probability density
# becomes very low after this point for a mean of 12.
x = np.linspace(0, 60, 500)

# Calculate the Probability Density Function (PDF) for each x-value
expon_pdf = vehicle.pdf(x)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x, expon_pdf, color='skyblue', linewidth=2)

# Add title and labels
plt.title(f'Exponential Probability Distribution (Mean = {lamda})')
plt.xlabel('Time Between Arrivals (seconds)', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)

# Display the plot
plt.show()
```
